START GAME

Show title "ROCK PAPER SCISSORS"
Wait a moment

Show rules:

Rock (r) beats Scissors (s)
Scissors (s) beats Paper (p)
Paper (p) beats Rock (r)
You can type full words or r/p/s
First to 3 wins

Wait a moment

SET play_again = "yes"

WHILE play_again == "yes":

SET user_score = 0
SET computer_score = 0
SET invalid_input = 0
SET round_number = 1

WHILE user_score < 3 AND computer_score < 3 AND invalid_input < 4:

Show round number

ASK user for choice (rock/paper/scissors or r/p/s)
Convert input to lowercase

IF input is in shortcuts:
→ Convert to full word (r→rock, p→paper, s→scissors)

IF user_choice is valid (rock/paper/scissors):

 Generate random computer choice

 Show both choices

 Show "calculating..." and wait a moment

 IF user_choice == computer_choice:
  Show "tie message"

 ELSE IF user wins (based on rules):
  Show win message
  Increase user_score by 1

 ELSE:
  Show lose message
  Increase computer_score by 1

 Increase round_number by 1

 Show current score

ELSE:
 Show invalid move message
 Increase invalid_input by 1

END INNER LOOP

AFTER GAME ENDS:

IF user_score == 3:
→ Show "You won the game"

ELSE IF computer_score == 3:
→ Show "You lost the game"

ELSE:
→ Show "Game ended unexpectedly"

ASK TO PLAY AGAIN (with 2-attempt limit):

SET attempts = 0

WHILE attempts < 2:

 ASK: "Play again? (yes/no or y/n)"

 IF input == "y" → play_again = "yes" → BREAK
 IF input == "n" → play_again = "no" → BREAK
 IF input == "yes" or "no" → ACCEPT → BREAK

 ELSE:
  Show error message
  Increase attempts

IF attempts == 2:
→ play_again = "no"

END OUTER LOOP (when play_again == "no")

Show goodbye message

END GAME

```python
import random
import time

choices = ["scissors", "rock", "paper"]
play_choices = ["yes", "no"]

rules = {
    "rock": "scissors",
    "paper": "rock",
    "scissors": "paper"
}

shortcuts = {
    "r": "rock",
    "p": "paper",
    "s": "scissors"
}


def decide_winner(user_choice, computer_choice):
    if user_choice == computer_choice:
        return "tie"
    elif computer_choice == rules[user_choice]:
        return "win"
    else:
        return "lose"


def game_intro():
    print("""
    
    *** ROCK PAPER SCISSORS ***
    """)
    time.sleep(1)

    print("""
    Rules:
    - Rock (r) beats scissors (s)
    - Scissors (s) beats paper (p)
    - Paper (p) beats rock (r)
    - You can type the full word or just r, p, or s.
    - First to 3 wins the game
    
    """)
    time.sleep(2)


def game_final_msg():
    print("**********************")
    print("\nThanks for playing!\n")
    print("See you next time 👋")


def calculating():
    print("Calculating result...")
    time.sleep(2)


def display_score(user_score, computer_score):
    print("**** Current Score ****")
    print(f"You: {user_score} | Computer: {computer_score}")


def get_play_again():
    for i in range(2):
        play_again = input("Do you want to play again? (Yes/No): ").strip().lower()

        if play_again == "y":
            return "yes"
        elif play_again == "n":
            return "no"
        elif play_again in play_choices:
            return play_again

        print("Please choose Yes, No, Y or N.")
        print()

    return "no"


def play_game():
    game_intro()

    play_again = "yes"

    while play_again == "yes":
        user_score = 0
        computer_score = 0
        invalid_input = 0
        round_number = 1

        while user_score < 3 and computer_score < 3 and invalid_input < 4:

            print(f"\n========= Round {round_number} =========")

            user_choice = input(
                "Choose rock (r), paper (p), or scissors (s): "
            ).strip().lower()

            if user_choice in shortcuts:
                user_choice = shortcuts[user_choice]

            if user_choice in choices:

                computer_choice = random.choice(choices)

                print(f"You chose: {user_choice}")
                print(f"Computer chose: {computer_choice}")

                calculating()

                result = decide_winner(user_choice, computer_choice)

                if result == "tie":
                    print("\n🤝 It's a tie!\n")

                elif result == "win":
                    print("\n🎉 You've won this round!\n")
                    user_score += 1

                else:
                    print("\n❌ Sorry, you've lost this round!\n")
                    computer_score += 1

                display_score(user_score, computer_score)
                round_number += 1

            else:
                print("\n🤔 That's not a valid move... try rock, paper, or scissors!\n")
                invalid_input += 1

        if user_score == 3:
            print("=========================================")
            print("\n🎉 Congratulations! You WON the game!\n")
            print("=========================================")

        elif computer_score == 3:
            print("=========================================")
            print("\n💀 Sorry! You LOST the game!\n")
            print("=========================================")

        else:
            print("=========================================")
            print("\n😅 Game over — the keyboard betrayed you!\n")
            print("=========================================")

        play_again = get_play_again()

    game_final_msg()
```

```python
play_game()
```
